```python
import pandas as pd
from functools import reduce

# Tip dataset
data = {
    'total_bill': [16.99, 10.34, 21.01, 23.68, 24.59],
    'tip': [1.01, 1.66, 3.50, 3.31, 3.61],
    'sex': ['Female', 'Male', 'Male', 'Male', 'Female'],
    'smoker': ['No', 'No', 'No', 'No', 'No'],
    'day': ['Sun', 'Sun', 'Sun', 'Sun', 'Sun'],
    'time': ['Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner'],
    'size': [2, 3, 3, 2, 4]
}

df = pd.DataFrame(data)
```

```python



# Example DataFrame


# Find the sum of the tip using reduce
suml = reduce(lambda a, b: a + b, df['tip'])
print(suml)
```

```python
#find out the sum of the total_bill of each day
sum_bill = df.groupby('day')['total_bill'].sum()
print("Sum of total bill for each day:")
print(sum_bill)
```

```python
#Create a list containing tip values greater than ‘5’ using ‘filter’ function
tips_gt_5 = list(filter(lambda x: x > 5, df['tip']))
print("\nTip values greater than 5:")
print(tips_gt_5)
```
