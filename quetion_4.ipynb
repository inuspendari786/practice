```python
import numpy as np
import pandas as pd
from functools import reduce

# Sample data
numbers = [1, 2, 3, 4, 5]

# i) map() → square each number
squares = list(map(lambda x: x**2, numbers))
print("Squares using map():", squares)

# ii) apply() → apply function to Pandas column
df = pd.DataFrame({'values': numbers})
df['double'] = df['values'].apply(lambda x: x * 2)
print("\nDataFrame with apply():\n", df)

# iii) reduce() → sum of all numbers
total = reduce(lambda x, y: x + y, numbers)
print("\nSum using reduce():", total)

# iv) lambda() → inline function for cube
cubes = list(map(lambda x: x**3, numbers))
print("\nCubes using lambda():", cubes)
```

```python
sum = reduce(lambda a,b:a+b , numbers)
print("\n sum using reduce():",sum)
```
