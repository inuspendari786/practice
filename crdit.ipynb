```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(5,5))
ax = fig.add_axes([0,0,1,1])

# Scatter plot
ax.scatter(df["Credit_Limit"], df["Avg_Purchase"])

# Titles and labels
ax.set_title("jhuyggk", fontsize=12)
ax.set_ylabel("Credit_Limit")
ax.set_xlabel("Avg_Purchase")

plt.show()
```

```python
import matplotlib.pyplot as plt

df_list = [df["Credit_Limit"], df["Avg_Purchase"]]

fig = plt.figure(figsize=(5,5))
ax = fig.add_axes([0,0,1,1])

# Boxplot with labels
ax.boxplot(df_list, widths=0.5)
ax.set_title("hjhhjkjjj")
ax.set_xticklabels(["Credit_Limit", "Avg_Purchase"])

plt.show()
```
