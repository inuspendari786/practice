```python
import numpy as np
```

```python
s_1=np.array([1,2,3,4,5,6,7])
s_2=np.array([6,7,8,9,10,11,12,13])

using_apply=np.array(list(map(lambda x,y:x+y, s_1,s_2)))
print(using_apply)
```

```python
from functools import reduce
s_1=np.array([1,2,3])
s_2=np.array([1,2,3])

using_apply=reduce(lambda x,y:x-y, s_1,s_2)
print(using_apply)
```
