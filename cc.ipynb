```python
import pandas as pd
from functools import reduce

# Student dataset
data = {
    'name': ['Arun', 'Bhavna', 'Chitra', 'Deepak', 'Esha'],
    'math': [85, 67, 92, 45, 78],
    'science': [75, 88, 69, 55, 90],
    'english': [80, 72, 95, 60, 85]
}

df = pd.DataFrame(data)


```

```python
total= df["math"].sum()
print("\n to_S_o")
print(total)
```

```python
eng= list(filter(lambda x :x >41,df["english"]))
print(eng)
```

```python
sum= reduce(lambda a,b:a+b,df["science"])
print(sum)
```
