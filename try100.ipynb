```python
import pandas as pd

A = pd.Series([20, 30, 40, 50, 60])
B = pd.Series([50, 60, 70, 80, 90])

# (i) Items not common to both
print("Not Common:", pd.concat([A, B]).drop_duplicates(keep=False))

# (ii) Smallest and largest element in A
print("Smallest:", A.min())
print("Largest:", A.max())

# (iii) Sum of Series B
print("Sum of B:", B.sum())

# (iv) Mean of Series A
print("Mean of A:", A.mean())

# (v) Median of Series B
print("Median of B:", B.median())
```
