```python
import pandas as pd
```

```python
df=pd.read_csv("credit.csv")
df.head()
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt 
fig=plt.figure(figsize=[5,5])
ax=fig.add_axes([0,0,1,1])

ax.scatter(df["Credit_Limit"],df["Avg_Purchase"])
ax.set_title("uhijh")
ax.set_ylabel("Credit_Limit")
ax.set_xlabel("Avg_Purchase")
plt.show()
```

```python
list=df["Credit_Limit"],df["Avg_Purchase"]
fig= plt.figure(figsize=[5,5])
ax=fig.add_axes([0,0,1,1])

ax.boxplot(list,widths=0.5)
ax.set_title("credit limit and avg purchuse")
ax.set_xticklabels(["Credit_Limit","Avg_Purchase"])
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count per Income_Category
list_1 = df.groupby("Income_Category").size().reset_index(name="count")

fig = plt.figure(figsize=(10,10))
ax = fig.add_axes([0,0,1,1])

# Bar chart
ax.bar(list_1["Income_Category"], list_1["count"])

# Titles and labels
ax.set_title("Clusters of Income Category")
ax.set_ylabel("Count")
ax.set_xlabel("Income_Category")

#plt.tight_layout()




#plt.show()
```
