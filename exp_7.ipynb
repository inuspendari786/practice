```python
import pandas as pd
```

```python
df=pd.read_csv("exp_7.csv")
print(df.head())
```

```python
df["city_mileage"]=df["mpg"]*0.75
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
mpg_list=[df["mpg"],df["city_mileage"]]
plt.boxplot(mpg_list)
plt.title("mpg vs city milege")
plt.ylabel("mpg")
#plt.xlabel([1,2],["Average mpg","city mpg"])
plt.show()
```

```python
min= df["mpg"].mean()
print(min)
```

```python
import seaborn as sns    
sns.scatterplot(data=df, x="horsepower",y="mpg")
plt.title("ffff")
plt.xlabel("mpg")
plt.ylabel("horsepower")
plt.show()
correlation = df["horsepower"].corr(df["mpg"])
print(f"correlation vbetween horsepower and mpg:{correlation :.3f}")
```
