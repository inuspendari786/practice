```python
import pandas as pd
from functools import reduce

# Tip dataset
data = {
    'total_bill': [16.99, 10.34, 21.01, 23.68, 24.59],
    'tip': [1.01, 1.66, 3.50, 3.31, 3.61],
    'sex': ['Female', 'Male', 'Male', 'Male', 'Female'],
    'smoker': ['No', 'No', 'No', 'No', 'No'],
    'day': ['Sun', 'Sun', 'Sun', 'Sun', 'Sun'],
    'time': ['Dinner', 'Dinner', 'Dinner', 'Dinner', 'Dinner'],
    'size': [2, 3, 3, 2, 4]
}

df = pd.DataFrame(data)

sum = df.groupby("day")["total_bill"].sum()
print ("sum of total bill on wich day")
print(sum)
```

```python
tip=list(filter(lambda x: x >5 , df["tip"]))
print("\n tip value greater than 5")
print(tip)
```

```python
tip=reduce(lambda x, y: x+y,df["tip"])
print(tip)
```

```python
horsepower_mean = df['tip'].mean()
print("\nMean horsepower:", horsepower_mean)
```
