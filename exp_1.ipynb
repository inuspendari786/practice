```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("auto_mpg.csv")

# i) Cars built in Japan having 6 cylinders
print(df[(df['origin'] == 'Japan') & (df['cylinders'] == 6)])

# ii) Number of cars manufactured each year
print(df['model_year'].value_counts().sort_index())

# iii) Cars built in year 72
print(df[df['model_year'] == 72])

# iv) Boxplot for average vs city mileage
df['city_mileage'] = df['mpg'] * 0.75
plt.boxplot([df['mpg'], df['city_mileage']], labels=['Average Mileage', 'City Mileage'])
plt.title("Mileage Comparison")
plt.ylabel("Miles per Gallon")
plt.show()

# v) Scatterplot for horsepower vs mileage
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
plt.scatter(df['horsepower'], df['mpg'], color='blue')
plt.title("Horsepower vs Mileage")
plt.xlabel("Horsepower")
plt.ylabel("Mileage (mpg)")
plt.show()
```
